```python
# Step 1: Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 2: Create dataset
data = {
    "Experience": [1, 2, 3, 4, 5, 6, 7, 8],
    "Salary": [30000, 35000, 40000, 45000, 50000, 55000, 60000, 65000]
}

df = pd.DataFrame(data)

# Step 3: Define independent and dependent variables
X = df[["Experience"]]
y = df["Salary"]

# Step 4: Train model
model = LinearRegression()
model.fit(X, y)

# Step 5: Predict salary
experience_input = np.array([[5]])
predicted_salary = model.predict(experience_input)

print("Predicted Salary for 5 years experience:", predicted_salary[0])

# Step 6: Plot graph
plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color='red')
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Salary Prediction Model")
plt.show()

```
